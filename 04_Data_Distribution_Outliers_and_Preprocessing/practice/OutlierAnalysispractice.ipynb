```python
#turkey method
def outlierdect(column):
    sorted(column)
    Q1,Q3=np.percentile(column,[25,75])
    IQR=Q3-Q1
    lowerRange = Q1-(1.5*IQR)
    upperRange = Q3+(1.5*IQR)
    return lowerRange,upperRange
```

```python
lr,ur = outlierdect(dataset['Monthly_Sales'])
print(lr,ur)
```

```python
processData=dataset[(dataset['Monthly_Sales'] >= lr)*(dataset['Monthly_Sales'] <= ur)]
```
