```python
#Handling Ordinal Data
# Ordinal Data is categorical data with ranking
# Dealing with ordinal data, you can use Ordinal Encoding
# Ordinal Encoding means
#    - Replace Category with its associated rank
#    - Replace Category with the index location of unique value
```

```python
import pandas as pd
```

```python
data2 = {
    'Satisfaction': ['Neutral','Satisfied','Very Satisfied','Dissatisfied', 'Very Dissatisfied','Neutral']
}
df = pd.DataFrame(data2)
df
```

```python
mapping = {"Very Dissatisfied": 0,
"Dissatisfied": 1,
           "Neutral": 2,
           "Satisfied": 3,
           "Very Satisfied": 4}

df['SatisfactionEncoded'] = df['Satisfaction'].map(mapping)
df
```
