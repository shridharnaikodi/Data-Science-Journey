```python
# Performing Hypothesis Testing
#
# 1. Create a Viable Question Goal (The question must result in a Binary Outcome)
# 2. Convert the question goal into Hypothesis (h0 and h1)
# 3. Select the Statistical Testing Formula to Validate the Hypothesis (Check who wins?)
# 4. Select/Determine SL of the Project
# 5. Calc the p-value using Statistical Testing formula and compare the same with the SL to check who wins
#
#
```

```python
# TEST FOR FEATURE ELIMINATION | Feature Comparison (In terms of Inferential Stats | Model Training | Model Building)
#
# Compare two column/variables to check if they are statistically similar
#
#                                     Test for Feature Elimination
#                                                |
#      -------------------------------------------------------------------------------------------
#      |                                                                                         |
#  Numerical and Numerical Pair                           Numerical and Categorical Pair | Categorical and Categorical
#       |
#   ----------------------------------------
#.  |                                      |
# if both pass Normality Test      if both fail Normality Test | if one of them fail normality
#   use Parametric test              use Non-parametric Test
#
#
# if col1 and col2 are Normally Dist:
#          Use Parametric Test to Identify whether to preserve or eliminate col
# else:
#          Use Non-Parametric Test to Identify whether to preserve or eliminate col
#
#
# Formulate the Hypothesis
#
# h0 --------- Eliminate one of the variable
# ha --------- Preserve both variable
```

```python
# Goal: Check if RDSPend and Admin Spend is Statistically Significant | similar | twins
#
# Step1: Check if RDSpend and AdminSpend is Normally Distributed using Shapiro test
```

```python
SL = 0.05

from scipy.stats import shapiro

stat, p = shapiro(data["R&D Spend"])

if p >= SL:
  print("ALternate Hypothesis (Ha) ---- RDSpend is Normally Dist")
else:
  print("Null Hypothesis (H0) ---- RDSpend is Not Normally Dist")
```

```python
SL = 0.05

from scipy.stats import shapiro

stat, p = shapiro(data["Administration"])

if p >= SL:
  print("ALternate Hypothesis (Ha) ---- Administration is Normally Dist")
else:
  print("Null Hypothesis (H0) ---- Administration is Not Normally Dist")
```

```python
# Goal: Check if RDSPend and Admin Spend is Statistically Significant | similar | twins

# 1. Create a viable question goal
#
# Check if RDSpend and Admin are Statistically Same
#
# 2. Convert the question into Hypothesis
#
#  H0 -------- RDSpend and AdminSpend are Statistically Same. Therefore eliminate one of them
#  H1 -------- RDSpend and Admin are Statistically Different. Therefore Preserve both of them
#
# 3. Select Test
#
#     Student T Test
#
# 4. Select SL
#
#      0.05
#
# 5. Calc p value

SL = 0.05

from scipy.stats import ttest_ind

stat, p = ttest_ind(data["R&D Spend"], data["Administration"])

if p <= SL:
  print("Alternate Hypothesis (H1) ---- RDSpend and Admin are Statistically Different. Therefore Preserve both of them")
else:
  print("Null Hypothesis (H0) --RDSpend and AdminSpend are Statistically Same. Therefore eliminate one of them")
```

# Chi-Square Test

```python
# Chi-square for Feature Elimination
#
# It is meant to deeal with Feature(independent Variable) and Label(dependent variable) pair
#
# 1. Feature is categorical and label is numerical
# 2. Feature is categorical and label is categorical
# 3. Feature is numerical and label is categorical
#
# The goal this test is to check whether there exists any kind of STATISTICAL RELATIONSHIP between two columns
```

```python
data.head()
```

```python
#Check if State and Profit have any Stat relationship
# Contigency Table | Pivot Table

ct = pd.crosstab(data["State"], data["Profit"])
ct
```

```python
# from scipy.stats import chi2_contingency
# ?chi2_contingency
```

```python
SL = 0.05

from scipy.stats import chi2_contingency

_,pvalue,_,_ = chi2_contingency(ct)

if pvalue <= SL:
  print("Alternate Hypothesis (H1) ---- State and Profit has some Statistical Relationship. Thus Preserve State column")
else:
  print("Null Hypothesis (H0)------ State and Profit has NO Statistical Relationship. Thus Eliminate State column")
```
