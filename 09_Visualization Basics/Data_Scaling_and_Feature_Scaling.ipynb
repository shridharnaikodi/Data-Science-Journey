```python
# Feature Preparation Technique
#
#                                 Scaling
#                                    |
#     -----------------------------------------------------------------------
#     |                                                                     |
#  Normal Distribution                                            Skew Distribution
#
#  1. StandardScaler                                              1. RobustScaler
#  2. MinMaxScaler  (maximum cols are discrete)
```

```python
#
# Feature
#
#. age.  ------------------ 18 - 60
#  salary ----------------- 20000 - 400000
#  experience ------------- 0 - 50
#
#
# Bring all feature cols in one scale to avoid any biasness(partiality based on huge numbers)
```

```python
#StandardScaler --It transforms the data such that the mean is zero and stddev is 1

import numpy as np

from sklearn.preprocessing import StandardScaler

data1 = np.array([[100],[200],[300],[400],[500]])

sc = StandardScaler()

transformedData = sc.fit_transform(data1)
```

```python
#MinMaxScaler ---- Its transforms the data in the defined range. Default range is 0,1

import numpy as np

from sklearn.preprocessing import MinMaxScaler

data1 = np.array([[100],[200],[300],[400],[500]])

sc = MinMaxScaler()

transformedData = sc.fit_transform(data1)
```

```python
transformedData
```

```python

import numpy as np

from sklearn.preprocessing import MinMaxScaler

data1 = np.array([[100],[200],[300],[400],[500]])

sc = MinMaxScaler( feature_range=(10,20) )

transformedData = sc.fit_transform(data1)

transformedData
```

```python
#RobustScaler ---- uses 1.5 IQR rule to define the scale range

# newValue = (currentValue - median) / iqr

import numpy as np

from sklearn.preprocessing import RobustScaler

data1 = np.array([[100],[200],[300],[400],[500]])

sc = RobustScaler()

transformedData = sc.fit_transform(data1)
transformedData
```
