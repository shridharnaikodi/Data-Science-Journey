```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.info()
```

```python
tips.head()
```

## Two types of columns (Broad Spectrum)

 1. Quantitative Data (Numerical Data)
       Continuous ND
       Discrete ND

 2. Qualitative Data (Non-numerical Data)
        Categorical Data
        Ordinal Data
        Pure String

## Dealing with Quantitative Data (Numerical Data)

1. Distribution Plot --- (KDE Plot, Histogram) --- Ideal purpose is to check  if the data is normally dist or
                                                    has some skewness
2. Box plot -------------------------------------- Checking if outlier exists or not

3. Scatterplot or Joint Plot --------------------- Correlation Analysis

```python
# 1. Distribution Plots

#Plotting histogram

sns.displot(tips['total_bill'])
```

```python
sns.displot(tips['total_bill'], kind="kde")
```

```python
#Pandas

tips['total_bill'].plot.kde()
```

```python
#Box Plot

sns.boxplot(x=tips['total_bill'])
```

```python
#Multi-variate analysis
sns.boxplot(tips)
```

```python
#Pandas

tips['total_bill'].plot.box()
```

```python
tips.plot.box()
```

```python
# Scatter plot

sns.scatterplot(x='total_bill', y='tip', data=tips)
```

```python
tips['total_bill'].corr(tips['tip'])
```

```python
#PANDAS

tips.plot.scatter(x='total_bill', y='tip')
```

```python
# Joint Plot

sns.jointplot(x='total_bill', y='tip', data=tips)
```

```python
# FacetGrid ----- Give categorical insights

g = sns.FacetGrid(tips, hue='sex') #hue is to add categorical insight
g.map(sns.scatterplot, 'total_bill', 'tip') #Create scatterplot
g.add_legend() #Labelled the categorical column
```

```python
#Heatmap

corrMatrix = tips.corr(numeric_only=True)
corrMatrix
```

```python
sns.heatmap(corrMatrix, annot=True)
```

## Non-Numerical Data

1. Pie Chart
2. Bar Plot
3. Count Plot

```python
sns.countplot(tips['day'])
```

```python
tips['time'].unique()
```

```python
sns.countplot(tips[tips['time'] == 'Lunch']['day'])
```

```python
sns.countplot(tips[tips['time'] == 'Dinner']['day'])
```

```python
sns.countplot(data=tips, x="day", hue="sex")
```

```python
sns.countplot(data=tips, x="day", hue="smoker")
```

```python
sns.countplot(data = tips[tips['time'] == 'Lunch'], x="day", hue="smoker")
```

```python
sns.countplot(data = tips[tips['time'] == 'Dinner'], x="day", hue="smoker")
```

```python
#Pie Chart --- Only pandas support pie chart

tips['day'].value_counts().plot.pie()
```
